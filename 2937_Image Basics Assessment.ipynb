### Import necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

### Open 'image2.jpg' from the data folder using open cv and display it in the notebook. Make sure to correct for the RGB order

```python
img = cv2.imread('image2.jpg')
img
```

##### Before RGB correction

```python
plt.imshow(img)
```

### After correction

```python
img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img1)
```

### Draw an empty RED rectangle around toms face and display the image in the notebook

```python
cv2.rectangle(img1, pt1 = (300, 420), pt2 = (600, 750), color = (255,0,0), thickness = 10)
```

```python
plt.imshow(img1)
```

### Draw a text named 'TOM' above toms face

```python
font = cv2.FONT_HERSHEY_TRIPLEX
cv2.putText(img1, text = 'TOM', org = (500,850), fontFace = font, fontScale = 4, color = (0,0,0), thickness = 5)
plt.imshow(img1)
```

### Do the same steps for 'Jerry'

```python
cv2.rectangle(img1, pt1 = (100,500), pt2 = (300,700), color = (255,0,0), thickness = 5)
plt.imshow(img1)
```

```python
font = cv2.FONT_HERSHEY_TRIPLEX
cv2.putText(img1, text = 'JERRY', org = (150,750),fontFace = font, fontScale = 2, color = (0,0,0), thickness = 5)
plt.imshow(img1)
```

### Convert the image to grayscale

```python
img_gray = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
plt.imshow(img_gray, cmap = 'gray')
```

### Display the image in a larger size

```python
plt.figure(figsize = (14,10))
plt.imshow(img_gray, cmap = 'gray')
```
