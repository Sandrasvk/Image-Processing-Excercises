### Import necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

### Read 'chess.jpg'

```python
def load_img():
    img = cv2.imread('chess.jpg').astype(np.float32) / 255
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img
```

### View the image with matplotlib

```python
def display_img(img):
    fig = plt.figure(figsize = (6,6))
    ax = fig.add_subplot(1,1,1)
    ax.imshow(img)

img = load_img()
display_img(img)
```

### Increase brightness by gamma correction

```python
img = load_img()
gamma = 1/8
effected_image = np.power(img,gamma)
display_img(effected_image)
```

### Apply Gaussian Blur on the original image

```python
img = load_img()
blurred_img = cv2.GaussianBlur(img,(5,5),10)
display_img(blurred_img)
```

### Apply median Blurring

```python
img = load_img()
median = cv2.medianBlur(img,5)
display_img(median)
```

### Apply bilateral Filtering

```python
img = load_img()
blur = cv2.bilateralFilter(img,7,60,60)
display_img(blur)
```
