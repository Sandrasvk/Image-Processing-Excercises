### Import necessary Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

### Read 'image1.jpg' with opencv

```python
img = cv2.imread('image1.jpg')
```

```python
img
```

### Plot the image with matplotlib

```python
image = cv2.imread('image1.jpg')
plt.imshow(image)
```

### Convert the image to RGB

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
```

### Check the shape of the image

```python
image_rgb.shape
```

### Resize the image to (400,300)

```python
img2 = cv2.resize(image_rgb, (400,300))
```

```python
plt.imshow(img2)
```

```python
img2.shape
```

### Resize the image by ratio. Use 0.5 height and width

```python
w_ratio = 0.5
h_ratio = 0.5
```

```python
new_img = cv2.resize(image_rgb,(0,0),img2, w_ratio, h_ratio)
```

```python
plt.imshow(new_img)
```

### Flip the image along central x axis

```python
new_img = cv2.flip(new_img,0)
plt.imshow(new_img)
```

### Flip the image along central y axis 

```python
new_img = cv2.flip(img2,1)
plt.imshow(new_img)
```
