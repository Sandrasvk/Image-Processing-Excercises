### Import opencv, numpy and matplotlib 

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Create a function that outputs an image array with some text

```python
def load_img():
    blank_img = np.zeros((600,600))
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img, text = 'EDURE', org = (50,300), fontFace = font, fontScale = 5, color = (255,255,255), thickness = 25, lineType = cv2.LINE_AA)
    return blank_img
```

### Create a function that takes image array as argument and displays the image in a larger size

```python
def display_img(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img,cmap='gray')
```

### Display the image created with the function

```python
img = load_img()
display_img(img)
```

### Erode away boundaries of the foreground objects

```python
img = load_img()
kernel = np.ones((5,5))
erosion = cv2.erode(img,kernel,iterations=3)

display_img(erosion)
```

### Create an image with backgrond noise

```python
img = load_img()
white_noise = np.random.randint(low=0,high=2,size=(600,600))
white_noise
```

```python
white_noise = white_noise * 255
```

```python
noise_img = img + white_noise
```

```python
display_img(noise_img)
```

### Remove the background noise with opening

```python
opening = cv2.morphologyEx(noise_img,cv2.MORPH_OPEN,kernel)
```

```python
display_img(opening)
```

### Apply morphological Gradient on the original image

```python
img = load_img()
gradient = cv2.morphologyEx(img,cv2.MORPH_GRADIENT,kernel)
```

```python
display_img(gradient)
```
